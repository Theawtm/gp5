```python
class MLP(nn.Module):
    def __init__(self):
        super(MLP, self).__init__()
        self.fc1 = nn.Linear(9, 64)
        self.act1 = nn.SiLU()
        self.fc2 = nn.Linear(64, 32)
        self.act2 = nn.SiLU()
        self.fc3 = nn.Linear(32, 2)

    def forward(self, x):
        x = self.act1(self.fc1(x))
        x = self.act2(self.fc2(x))
        x = self.fc3(x)
        return x
```
